```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv(r"D:\Data Analyst\Projects\BMW Sales Portfolio\1. Dataset\BMW sales data (2010-2024).csv")

df.head()
```

```python
df
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
df.describe()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
print("Total Records:", len(df))

print("Total Sales Volume:", df["Sales_Volume"].sum())

print("Average Price:", df["Price_USD"].mean())

print("Average Mileage:", df["Mileage_KM"].mean())

print("Total BMW Models:", df["Model"].nunique())

print("Total Regions:", df["Region"].nunique())
```

```python
model_sales = df.groupby("Model")["Sales_Volume"].sum()

print(model_sales.sort_values(ascending=False))

print("\nTop 5 Models:")
print(model_sales.nlargest(5))
```

```python
region_sales = df.groupby("Region")["Sales_Volume"].sum()

print(region_sales.sort_values(ascending=False))
```

```python
fuel_sales = df.groupby("Fuel_Type")["Sales_Volume"].sum()

print(fuel_sales.sort_values(ascending=False))
```

```python
year_sales = df.groupby("Year")["Sales_Volume"].sum()

print(year_sales)
```

```python
plt.figure(figsize=(10,5))

year_sales.plot(kind="line", marker="o")

plt.title("BMW Year-wise Sales Trend")
plt.xlabel("Year")
plt.ylabel("Sales Volume")

plt.grid()
plt.show()
```

```python
model_sales.to_csv("BMW_Model_Analysis.csv")

region_sales.to_csv("BMW_Region_Analysis.csv")

fuel_sales.to_csv("BMW_Fuel_Analysis.csv")

print("Python Analysis Completed Successfully!")
```
